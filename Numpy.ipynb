**NUMPY**

NumPy is a powerful open-source Python library used for numerical computing.

Supports:
Multidimensional arrays
Mathematical operations on arrays
Linear algebra
Fourier transforms
Random number generation

Why is NumPy faster than Python lists?

Uses optimized C libraries internally.

```python
import numpy as np

arr1 = [1, 2, 3, 4]
print(arr1)

arr2 = [[1, 2, 3, 4], [5, 6, 7, 8]]
print(arr2)

arr3 = np.array([1, 2, 3])
print(arr3)

arr4 = np.arange(0, 20, 2)
print(arr4)
```

```python
#Zeros Array
zero = np.zeros((3, 3))
print(zero)
```

```python
#Ones Array
one = np.ones((3, 3))
print(one)
```

```python
#Identity Matrix
identity = np.eye(3)
print(identity)
```

```python
#Linspace
ls = np.linspace(2, 10, 5)
print(ls)
```

```python
#Random Numbers
randnum = (np.random.rand(2, 3) * 10).astype(int)
print(randnum)
```

```python
#Sum, Mean and Standard Deviation
array1 = [[1, 2, 3, 4],
          [5, 6, 7, 8]]

array2 = [[4, 6, 8, 3],
          [2, 5, 7, 9]]

add = np.sum(array2)
print(add)

mean = np.mean(array1)
print(mean)

std = np.std(array1)
print(std)
```

```python
#Minimum, Maximum, Argmin and Argmax
Min = np.min(array1)
print(Min)

Max = np.max(array1)
print(Max)

ArgMin = np.argmin(array1)
print(ArgMin)

ArgMax = np.argmax(array1)
print(ArgMax)
```

```python
#Matrix Multiplication
x = np.array([[3, 4],
              [5, 6]])

y = np.array([[8, 9],
              [6, 7]])

dot = np.dot(x, y)
print(dot)
```

```python
#Sorting
sorts = np.sort(array2)
print(sorts)
```
