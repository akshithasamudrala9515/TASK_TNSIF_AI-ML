1. What is EDA? How does it help?

Exploratory Data Analysis (EDA) is the process of analyzing and summarizing a cleaned dataset to understand its patterns, distribution, relationships, and characteristics before building a Machine Learning model.

How EDA helps:
Understand the structure of data.
Detect missing values and outliers.
Identify patterns and relationships.
Select useful features for modeling.
Improve model accuracy.
Types of EDA:
Univariate Analysis – Analysis of one variable.
Bivariate Analysis – Analysis of two variables.
Multivariate Analysis – Analysis of more than two variables.


2. What are Outliers? Explain the procedure to detect outliers.

Outliers are values that are unusually high or unusually low compared to the rest of the data.

Causes:
Human data entry mistakes
Measurement errors
Experimental errors
Natural variations
Fraudulent activities
Procedure (IQR Method):
Arrange the data in ascending order.
Find Q1 (25%) and Q3 (75%).
Calculate IQR = Q3 − Q1.
Lower Limit = Q1 − 1.5 × IQR
Upper Limit = Q3 + 1.5 × IQR
Values below the lower limit or above the upper limit are outliers.
Handle outliers by removing or replacing them (mean/median).

```python
import pandas as pd
import plotly.express as px

# Create dataset
data = {
    "Student": ["A", "B", "C", "D", "E"],
    "Marks": [10, 20, 30, 40, 50]
}

df = pd.DataFrame(data)

# Plot bar graph
fig = px.bar(df, x="Student", y="Marks",
             title="Student Marks")

fig.show()
```
