**DATASTRUCTURES**

**Stack**
Definition

A Stack is a linear data structure that follows the LIFO (Last In, First Out) principle.

The last element inserted is the first element removed.

Example: A stack of plates.

```python
#Stack using Class
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if self.is_empty():
            print("Stack is empty")
            return None
        return self.items.pop()

    def peek(self):
        if self.is_empty():
            print("Stack is empty")
            return None
        return self.items[-1]

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)

    def display(self):
        print(self.items)


stack = Stack()

stack.push(10)
stack.push(20)
stack.push(30)

stack.display()
print("Top:", stack.peek())

print("Popped:", stack.pop())

stack.display()
```

**Queue**

Queue follows FIFO (First In First Out)

Operations:
enqueue()
dequeue()
size()
front()

Applications:
CPU Scheduling
Print Queue
Task Processing
BFS (Breadth First Search)

```python
#Queue using Class
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, item):
        self.items.append(item)

    def dequeue(self):
        if self.is_empty():
            print("Queue is empty")
            return None
        return self.items.pop(0)

    def front(self):
        if self.is_empty():
            print("Queue is empty")
            return None
        return self.items[0]

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)

    def display(self):
        print(self.items)


q = Queue()

q.enqueue(10)
q.enqueue(20)
q.enqueue(30)

q.display()
print("Front:", q.front())

print("Dequeued:", q.dequeue())

q.display()
```

**Linked List**

A Linked List is a collection of nodes connected by links (references).

Each node contains:

Data
Reference (Pointer)

Types of Linked List:
Singly Linked List
Doubly Linked List
Circular Linked List

```python
#linkedlist
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def insert(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
        else:
            temp = self.head
            while temp.next:
                temp = temp.next
            temp.next = new_node

    def display(self):
        temp = self.head
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")


ll = LinkedList()

ll.insert(10)
ll.insert(20)
ll.insert(30)

ll.display()
```

**Trees**

A Tree is a hierarchical data structure made up of nodes connected by edges.

Applications of Trees
File Systems
Database Indexing
HTML DOM Structure
Organization Charts
Decision Trees in Machine Learning
Expression Trees in Compilers

```python
#Binary Tree
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


root = Node(10)
root.left = Node(20)
root.right = Node(30)

print("Root:", root.data)
print("Left Child:", root.left.data)
print("Right Child:", root.right.data)
```

```python
#Binary Tree Traversal (Inorder)
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def inorder(root):
    if root:
        inorder(root.left)
        print(root.data, end=" ")
        inorder(root.right)


root = Node(10)
root.left = Node(20)
root.right = Node(30)
root.left.left = Node(40)
root.left.right = Node(50)

print("Inorder Traversal:")
inorder(root)
```

```python
#Binary Tree Traversal (Preorder)
def preorder(root):
    if root:
        print(root.data, end=" ")
        preorder(root.left)
        preorder(root.right)


preorder(root)
```

```python
#Binary Tree Traversal (Postorder)
def postorder(root):
    if root:
        postorder(root.left)
        postorder(root.right)
        print(root.data, end=" ")


postorder(root)
```

```python
#Student Report Card
students = []

name = input("Enter your name: ")
age = int(input("Enter your age: "))

subjects = set()
num_subjects = int(input("Enter number of subjects: "))

students.append(name)

student_info = (name, age)
marks = {}

for i in range(num_subjects):
    subject = input("Enter subject name: ")
    mark = int(input("Enter marks: "))

    subjects.add(subject)
    marks[subject] = mark

total = sum(marks.values())
avg = total / num_subjects

print("\n****Student Report Card****")
print("Name:", student_info[0])
print("Age:", student_info[1])
print("Subjects:", subjects)

print("Marks:")
for subject, mark in marks.items():
    print(subject, ":", mark)

print("Total Marks:", total)
print("Average Marks:", avg)
```
