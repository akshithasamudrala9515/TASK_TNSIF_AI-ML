Introduction

2 data structures:

Series
A Series is a 1D array of indexed data. It is an object built on top of a numpy array.


```python
import numpy as np
import pandas as pd

s = pd.Series([1, 2, 3, 4, 5, 6])
print(s)
```

Inbuilt functions:

s.mean()
s.std()
s.min()
s.max()

Descriptive Statistics
count, mean, std, min, max, quartile values

```python
print(s.describe())
print(s.sum())
print(s.mean())
print(s.std())
print(s.min())
print(s.max())
```

Data Manipulation

s.map(func) – Applies a function to every element in the series (only for Series)
s.apply(func) – Similar to map, but more flexible (works for both Series & DataFrame)
s.sort_values() – Sort series
s.drop(labels) – Drop specified labels from series (not deleted permanently, returns new object)

```python
# Example
s.map(lambda x: x * 2)
```

```python
applies = s.apply(lambda x: x ** 2)
print(applies)
```

```python
s1 = pd.Series([4, 7, 54, 23, 92, 41])

sorted = s1.sort_values()
print(sorted)
```

```python
dropped = s.drop(1)
print(dropped)
```

Handling Missing Data

s.isnull() → Check for missing values, returns a Series of booleans



```python
s2 = pd.Series([2, 4, 6, np.nan, 7, 0])
print(s2.isnull())
```

s.notnull() → Opposite of isnull()

```python
print(s2.notnull())
```

s.fillna(value) → Fill missing value with specified value

```python
print(s2.fillna(0))
```

s.dropna() → Drops all rows that contain missing values (like None, NaN also drops)



```python
s3 = pd.Series([4, 7, 3, None, 8, 2])
print(s3.dropna())
```

**Indexing, Slicing & Filtering**

s.iloc[] – Purely integer-location based indexing.
(+) → Left to Right, (–) → Right to Left

```python
print(s.iloc[2])
print(s.iloc[-2])
```

s.loc[] – Label based indexing (only +ve, not -ve)

```python
print(s.loc[2])
print(s.loc[1:2])
```

s[s > n] – Filters & returns elements greater than "n"


```python
greater_num = s[s > 4]
print(greater_num)
```

**Aggregation**

s.sum() – Sums up the values
s.cumsum() – Cumulative sum

```python
print(s.sum())          # 21
print(s.cumsum())
```

s.aggregate(func) – Aggregate using one or more operations



```python
print(s.aggregate(['sum', 'min', 'max', 'mean']))
```
