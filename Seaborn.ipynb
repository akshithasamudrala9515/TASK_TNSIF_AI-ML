**Seaborn**

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'students': ['A', 'B', 'C', 'D'],
    'marks': [78, 85, 92, 88]
})

sns.barplot(x='students', y='marks', data=df)
plt.title("Students Marks")
plt.show()
```

```python
# Load built-in sample datasets
tips = sns.load_dataset("tips")
```

```python
#1. sns.scatterplot() — scatter plots
df = pd.DataFrame({
    'height': [120, 150, 180, 170],
    'weight': [50, 45, 63, 71]
})

sns.scatterplot(x='height', y='weight', data=df)
plt.show()
```

```python
#2. sns.lineplot() — line plots
df = pd.DataFrame({'day': [1, 2, 3, 4], 'sales': [100, 150, 130, 170]})
sns.lineplot(x='day', y='sales', data=df)
plt.show()
```

```python
#sns.barplot()
sns.barplot(x='day', y='sales', data=df)
plt.show()
```

```python
#4. sns.countplot()
tips = sns.load_dataset("tips")
sns.countplot(x='day', data=tips)
plt.show()
```

```python
#5. sns.boxplot()
sns.boxplot(x='day', y='total_bill', data=tips)
plt.show()
```

```python
#6. sns.violinplot()
sns.violinplot(x='day', y='total_bill', data=tips)
plt.show()
```

```python
#7. sns.heatmap()
corr = tips.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.show()
```

```python
#8. sns.pairplot()
sns.pairplot(tips)
plt.show()
```

```python
#9. sns.distplot() / sns.histplot()
sns.histplot(tips['total_bill'], kde=True)
plt.show()
```

```python
#10. sns.regplot()
sns.regplot(x='total_bill', y='tip', data=tips)
plt.show()
```

```python
#11. sns.catplot()
sns.catplot(x='day', y='total_bill', data=tips, kind='box')
plt.show()
```
