**FUNCTIONS**

```python
#Function Without Parameters and Without Return Value
def display():
    print("Hello")

display()
```

```python
#Function With Parameters
def greet(name):
    print("Hello", name)

greet("Nikitha")
greet("Akshitha")
```

```python
# Function With Return Value
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
```

```python
# Function With Parameters and Return Value
def square(num):
    return num * num

print(square(5))
```

**Lambda Function**

```python
add = lambda a, b: a + b

print(add(2, 3))
```

```python
name = lambda a: a

print(name("Akshitha"))
```

```python
n = lambda: "ABC"

print(n())
```

```python
#**Conditional Statements in Functions**
def check(num):
    if num > 10:
        return "Welcome"
    else:
        return "Invalid"

print(check(15))
print(check(5))
```

```python
#Recursive Function (Factorial)
def fact(n):
    if n == 1:
        return 1
    return n * fact(n - 1)

print(fact(8))
```
