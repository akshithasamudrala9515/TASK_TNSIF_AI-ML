**TASK**
Spam Mail Detection

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression

# -----------------------------------
# Dataset
# -----------------------------------
df = pd.DataFrame({
    "Email": [
        "Win a brand new laptop today",
        "Your bank account has been credited",
        "Exclusive offer! Buy one get one free",
        "Team meeting scheduled for Monday",
        "Submit your assignment before 5 PM",
        "Earn money from home easily",
        "Your interview is confirmed",
        "Click here to receive your gift card",
        "Project presentation is tomorrow",
        "Congratulations! You are selected for a cash reward",
        "Monthly salary has been processed",
        "Limited time discount on all products"
    ],
    "Label": [
        "Spam",
        "Not Spam",
        "Spam",
        "Not Spam",
        "Not Spam",
        "Spam",
        "Not Spam",
        "Spam",
        "Not Spam",
        "Spam",
        "Not Spam",
        "Spam"
    ]
})

# -----------------------------------
# Independent and Dependent Variables
# -----------------------------------
X = df["Email"]
y = df["Label"]

# -----------------------------------
# Convert text into numerical features
# -----------------------------------
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(X)

# -----------------------------------
# Create and Train Model
# -----------------------------------
model = LogisticRegression()
model.fit(X, y)

# -----------------------------------
# Test Emails
# -----------------------------------
test_emails = [
    "Congratulations! You have won a free gift",
    "Please attend today's project meeting",
    "Claim your reward before it expires",
    "Assignment submission is due tomorrow",
    "Exclusive cash bonus waiting for you"
]

# Convert test emails
test_vectors = vectorizer.transform(test_emails)

# Predict
predictions = model.predict(test_vectors)

# Display Results
print("Email Predictions:\n")

for email, label in zip(test_emails, predictions):
    print(f"Email: {email}")
    print(f"Prediction: {label}\n")

# -----------------------------------
# Visualization
# -----------------------------------
df["Label"].value_counts().plot(
    kind="bar",
    color=["green", "red"]
)

plt.title("Spam vs Not Spam Emails")
plt.xlabel("Email Category")
plt.ylabel("Number of Emails")
plt.grid(axis="y")
plt.show()
```
