```python
# ==========================================
# Rainfall Dataset Cleaning
# ==========================================

# Import Libraries
import pandas as pd
import numpy as np
from google.colab import files

# Upload CSV File
uploaded = files.upload()

# Load Dataset
df = pd.read_csv("rainfall.csv")

# Display First 5 Rows
print("First 5 Rows:")
print(df.head())

# Dataset Information
print("\nDataset Info:")
print(df.info())

# ==========================================
# Convert Date Column
# ==========================================

df['date'] = pd.to_datetime(df['date'], errors='coerce')

# ==========================================
# Handle Missing Values
# ==========================================

print("\nMissing Values Before Cleaning:")
print(df.isnull().sum())

# Fill numerical missing values with median
num_cols = ['rainfall', 'temperature', 'humidity', 'wind_speed']

for col in num_cols:
    if col in df.columns:
        df[col].fillna(df[col].median(), inplace=True)

# Fill categorical missing values with mode
df['weather_condition'].fillna(df['weather_condition'].mode()[0], inplace=True)

# Drop rows where date is missing
df.dropna(subset=['date'], inplace=True)

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

# ==========================================
# Fix Text Inconsistencies
# ==========================================

df['weather_condition'] = (
    df['weather_condition']
    .str.strip()
    .str.lower()
    .str.capitalize()
)

print("\nUnique Weather Conditions:")
print(df['weather_condition'].unique())

# ==========================================
# Remove Duplicate Rows
# ==========================================

print("\nDuplicate Rows:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

# ==========================================
# Convert Data Types
# ==========================================

df['rainfall'] = pd.to_numeric(df['rainfall'], errors='coerce')
df['temperature'] = pd.to_numeric(df['temperature'], errors='coerce')
df['humidity'] = pd.to_numeric(df['humidity'], errors='coerce')
df['wind_speed'] = pd.to_numeric(df['wind_speed'], errors='coerce')

# ==========================================
# Detect and Remove Outliers (IQR Method)
# ==========================================

for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower) & (df[col] <= upper)]

print("\nDataset Shape After Removing Outliers:")
print(df.shape)

# ==========================================
# Final Clean Dataset
# ==========================================

print("\nFinal Dataset:")
print(df.head())

print("\nFinal Dataset Info:")
print(df.info())

# ==========================================
# Save Cleaned Dataset
# ==========================================

df.to_csv("cleaned_rainfall.csv", index=False)

print("\nCleaned dataset saved as 'cleaned_rainfall.csv'")

# Download the cleaned file
files.download("cleaned_rainfall.csv")
```
