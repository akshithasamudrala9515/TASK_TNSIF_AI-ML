**Supervised Learning**
supervised Learning is a type of Machine Learning in which the model learns from labeled data.

Every training example contains:

Input (Features / Independent Variables)
Output (Target / Dependent Variable)

formula
Input (X) → ML Algorithm → Output (Y)

workflow
Collect Data
      ↓
Label the Data
      ↓
Split into Training & Testing Data
      ↓
Train the Model
      ↓
Evaluate the Model
      ↓
Predict New Data

Types of Supervised Learning
Regression
Classification

**Regression**

Regression is used when the output variable is continuous (numerical).

It predicts continuous numerical values.

**Types of Regression**
Simple Linear Regression
Multiple Linear Regression
Polynomial Regression
Decision Tree Regression
Random Forest Regression
Support Vector Regression (SVR)

**1.Simple Linear Regression**
formula:y = mx + c

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
df = pd.DataFrame({
    "Marks": [50, 40, 35, 60, 70, 90],
    "Rank": [8, 9, 11, 6, 4, 2]
})

X = df[["Marks"]]
y = df["Rank"]

model = LinearRegression()
model.fit(X, y)

prediction = model.predict([[80]])

print("Predicted Rank:", prediction[0])
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

**2.Multiple Linear Regression**
Formula:
y = b0 + b1x1 + b2x2 + ... + bnxn

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Create dataset
df = pd.DataFrame({
    "Experience": [1, 2, 3, 4, 5],
    "Education": [12, 14, 16, 18, 20],
    "Salary": [25000, 30000, 40000, 50000, 60000]
})

# Independent variables
X = df[["Experience", "Education"]]

# Dependent variable
y = df["Salary"]

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Predict salary for a person with:
# Experience = 6 years
# Education = 18 years
prediction = model.predict([[6, 18]])

print("Predicted Salary:", prediction[0])
print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
```

**3.decision tree**

```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor

# Create dataset
df = pd.DataFrame({
    "Experience": [1, 2, 3, 4, 5, 6],
    "Salary": [25000, 30000, 40000, 50000, 60000, 70000]
})

# Independent variable
X = df[["Experience"]]

# Dependent variable
y = df["Salary"]

# Create model
model = DecisionTreeRegressor(random_state=42)

# Train model
model.fit(X, y)

# Predict salary for 4.5 years of experience
prediction = model.predict([[4.5]])

print("Predicted Salary:", prediction[0])
```

4.polynomial regression

```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# Create dataset
df = pd.DataFrame({
    "Experience": [1, 2, 3, 4, 5, 6],
    "Salary": [25000, 30000, 45000, 65000, 90000, 120000]
})

# Independent variable
X = df[["Experience"]]

# Dependent variable
y = df["Salary"]

# Convert to polynomial features
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

# Train the model
model = LinearRegression()
model.fit(X_poly, y)

# Predict salary for 7 years of experience
prediction = model.predict(poly.transform([[7]]))

print("Predicted Salary:", prediction[0])
```

5.random forest

```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor

# Create dataset
df = pd.DataFrame({
    "Experience": [1, 2, 3, 4, 5, 6],
    "Salary": [25000, 30000, 45000, 60000, 80000, 100000]
})

# Independent variable
X = df[["Experience"]]

# Dependent variable
y = df["Salary"]

# Create model
model = RandomForestRegressor(n_estimators=100, random_state=42)

# Train model
model.fit(X, y)

# Predict salary for 7 years of experience
prediction = model.predict([[7]])

print("Predicted Salary:", prediction[0])
```

**Classification**

Classification predicts categories or classes, not numerical values.

The output belongs to predefined categories.

Examples
Yes / No
Pass / Fail
Spam / Not Spam
Fraud / Not Fraud

1.Logistic Regression

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression

df = pd.DataFrame({
    "Distance": [300, 200, 100, 50, 60],
    "Arrival": [0, 0, 1, 1, 1]
})

X = df[["Distance"]]
y = df["Arrival"]

model = LogisticRegression()
model.fit(X, y)

prediction = model.predict([[150]])

print("Prediction:", prediction[0])

if prediction[0] == 1:
    print("Arrived")
else:
    print("Not Arrived")
```

**2.Naive Bayes**

Naive Bayes is a supervised learning algorithm based on Bayes' Theorem.

It predicts the class having the highest probability.

Formula:
P(A|B) = (P(B|A) × P(A)) / P(B)

```python
import pandas as pd
from sklearn.naive_bayes import GaussianNB

# Create dataset
df = pd.DataFrame({
    "Message_Length": [10, 20, 30, 40, 50, 60],
    "Spam": [0, 0, 0, 1, 1, 1]
})

# Independent variable (Feature)
X = df[["Message_Length"]]

# Dependent variable (Target)
y = df["Spam"]

# Create model
model = GaussianNB()

# Train model
model.fit(X, y)

# Predict
prediction = model.predict([[45]])

print("Prediction:", prediction[0])

if prediction[0] == 1:
    print("Spam")
else:
    print("Not Spam")
```

3.decision tree

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

df = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5, 6],
    "Pass": [0, 0, 0, 1, 1, 1]
})

X = df[["Study_Hours"]]
y = df["Pass"]

model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

prediction = model.predict([[5]])

print("Prediction:", prediction[0])

if prediction[0] == 1:
    print("Pass")
else:
    print("Fail")
```

4.Random Forest Classifier


```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

df = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5, 6],
    "Pass": [0, 0, 0, 1, 1, 1]
})

X = df[["Study_Hours"]]
y = df["Pass"]

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X, y)

prediction = model.predict([[5]])

print("Prediction:", prediction[0])

if prediction[0] == 1:
    print("Pass")
else:
    print("Fail")
```

5.Support Vector Machine

```python
import pandas as pd
from sklearn.svm import SVC

df = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5, 6],
    "Pass": [0, 0, 0, 1, 1, 1]
})

X = df[["Study_Hours"]]
y = df["Pass"]

model = SVC(kernel='linear')
model.fit(X, y)

prediction = model.predict([[5]])

print("Prediction:", prediction[0])

if prediction[0] == 1:
    print("Pass")
else:
    print("Fail")
```

6.K-Nearest Neighbors (KNN)



```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier

# Create dataset
df = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5, 6],
    "Pass": [0, 0, 0, 1, 1, 1]
})

X = df[["Study_Hours"]]
y = df["Pass"]

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)

prediction = model.predict([[5]])

print("Prediction:", prediction[0])

if prediction[0] == 1:
    print("Pass")
else:
    print("Fail")
```
