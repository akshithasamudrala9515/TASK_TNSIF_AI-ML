```python
#class and object
def check_number(num):
    if num > 10:
        print("Welcome")
    else:
        print("Invalid")

check_number(15)
check_number(5)
```

```python
#employee class
class Emp:
    eid = 101
    ename = "Akshitha"

obj = Emp()

print("Employee Id:", obj.eid)
print("Employee Name:", obj.ename)
```

```python
#Constructor example
class Stu:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Stu("Akshitha", 22)

print("Name:", s1.name)
print("Age:", s1.age)
```

```python
#course class
class Course:
    def __init__(self, domain, duration):
        self.domain = domain
        self.duration = duration

s1 = Course("AIML", "3 months")

print("Domain:", s1.domain)
print("Duration:", s1.duration)
```

```python
#Student Class with Method
class Student:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Name of the student:", self.name)

s1 = Student("Akshitha")
s1.display()
```

```python
#1.Function to Add Two Numbers
def add(a, b):
    return a + b

result = add(2, 3)
print(result)
```

```python
#2.Recursive Factorial Function
def fact(n):
    if n == 1:
        return 1
    return n * fact(n - 1)

print(fact(9))
```

```python
#3.Student Class
class Student:
    name = "Akshitha"
    rno = 13
    marks = 90

    def display(self):
        print("Name:", self.name)
        print("Roll No:", self.rno)
        print("Marks:", self.marks)

s1 = Student()
s1.display()
```

```python
#4.Rectangle Class
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

r = Rectangle(15, 12)

print("Area:", r.area())
print("Perimeter:", r.perimeter())
```

```python
#Bank Account Class
class BankAccount:
    def __init__(self, acc_no, balance):
        self.acc_no = acc_no
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Balance:", self.balance)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawal Successful")
            print("Balance:", self.balance)
        else:
            print("Invalid Amount")

b = BankAccount(101, 5000)

b.deposit(1000)
b.withdraw(2000)
```
