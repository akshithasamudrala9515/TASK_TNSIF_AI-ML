**Python Data Types **


1. List ([])

Features:

Ordered
Mutable (can be changed)
Allows duplicates

```python
students = ["Ravi", "Sita", "Ravi"]

print(students)

students.append("Ram")
print(students)

students.insert(1, "Kiran")
print(students)

students.remove("Ravi")
print(students)

students.pop(1)
print(students)

students.sort()
print(students)

students.reverse()
print(students)

print("Length =", len(students))
print("Count of Ram =", students.count("Ram"))
```

2. Tuple (())

Features:

Ordered
Immutable (cannot be changed)
Allows duplicates

```python
marks = (80, 90, 70, 90, 85)

print(marks)

print("Length =", len(marks))

print("Count of 90 =", marks.count(90))

print("Index of 70 =", marks.index(70))
```

3. Set ({})

Features:

Unordered
Mutable
No duplicates

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

A.add(7)
print("Add:", A)

A.remove(2)
print("Remove:", A)

print("Union:", A.union(B))

print("Intersection:", A.intersection(B))

print("Difference:", A.difference(B))
```

4. Dictionary ({key:value})

Features:

Stores data in key-value pairs
Mutable
Keys must be unique

```python
student = {
    "name": "Akshitha",
    "age": 20,
    "course": "AIML"
}

print(student)

print(student.keys())

print(student.values())

print(student.items())

print(student.get("name"))

student.update({"name": "Hello"})
print(student)

student.pop("age")
print(student)
```

**STRINGS**

```python
#string conversions
s = "Python Programming"
print(s.upper())
print(s.lower())
print(s.capitalize())
print(s.title())
```

```python
#strip()
s = "   Python   "
print(len(s))

s = s.strip()
print(len(s))
print(s)
```

```python
s = "Python Programming"
print(s.replace("Python", "Java"))
```

```python
s = "Python Programming Language"
print(s.split())
```

```python
s = "Python Programming"
print(s.find("P"))
print(s.find("Programming"))
```

```python
s = "Python Programming"
print(s.count("m"))
print(s.startswith("Python"))
print(s.endswith("Programming"))
```

```python
#String Indexing
s = "Python"

print(s[0])    # First character
print(s[1])    # Second character
print(s[-1])   # Last character
```

```python
#Slicing
s = "Python Programming"

print(s[0:6])
print(s[7:])
print(s[:6])
```

```python
s = "Python"

print(s[0:6:2])
```
