```python
#Histogram
age = [18, 19, 20, 21, 22, 23, 24, 25, 25, 25]

plt.hist(age, bins=5)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
#Scatterplot
hrs = [1, 2, 3, 4, 5, 6]
marks = [30, 40, 50, 60, 75, 80]

plt.scatter(hrs, marks)
plt.title("Study Hours vs Marks")
plt.xlabel("hrs")
plt.ylabel("marks")
plt.show()
```

```python
#Piechart
labels = ['Python', 'Java', 'C++', 'JS']
values = [40, 25, 20, 15]

plt.pie(values, labels=labels, autopct='%.1f%%')
plt.title("Programming Lang Usage")
plt.show()
```

```python
#histogram on healthcare
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# Upload the dataset file
uploaded = files.upload()   # opens a file picker, choose your Kaggle CSV

# Get the uploaded filename automatically
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

# Take first 500 patients
ages = df["Age"].head(500)

# Visualize distribution
plt.figure(figsize=(8,5))
plt.hist(ages, bins=15, color="skyblue", edgecolor="black")
plt.title("Age Distribution of 500 Patients")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.show()

# Most common age gap (bin with highest count)
counts, bin_edges = pd.cut(ages, bins=10, retbins=True)
mode_bin = counts.value_counts().idxmax()
print("Most common age range:", mode_bin)
```

```python
#Student Marks Analysis

students = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J']
marks = [78, 85, 92, 67, 88, 95, 72, 81, 90, 76]

# Task 1: Display students' marks graphically
plt.bar(students, marks, color="skyblue")
plt.title("Student Marks Analysis")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()

# Task 2: Find the highest score
highest_score = max(marks)
topper = students[marks.index(highest_score)]
print("Highest Score:", highest_score, "by Student", topper)

# Task 3: Title & axis labels (already added above in Task 1)

# Task 4: Display marks in descending order & plot again
# Sort students and marks together, based on marks descending
sorted_data = sorted(zip(marks, students), reverse=True)
sorted_marks = [m for m, s in sorted_data]
sorted_students = [s for m, s in sorted_data]

plt.bar(sorted_students, sorted_marks, color="orange")
plt.title("Student Marks (Descending Order)")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```
