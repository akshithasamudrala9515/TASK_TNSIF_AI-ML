Conditional Statements in Python





Conditional statements are used to make decisions in a program based on certain conditions.

if Statement

Syntax:
Python
if condition:
    # code to execute if condition is True

```python
age = 18

if age >= 18:
    print("You are eligible to vote.")
```

if-else Statement

Syntax:

if condition:
    # code if condition is True
else:
    # code if condition is False

```python
num = 10

if num % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")
```

if-elif-else Statement
Syntax:
Python
if condition1:
    # code block
elif condition2:
    # code block
else:
    # code block

```python
marks = 75

if marks >= 90:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

Nested if Statement

Syntax:
Python
if condition1:
    if condition2:
        # code block

```python
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry Allowed")
```

Control statements


For Loop in Python
A for loop is used to repeat a block of code a specific number of times.


Syntax

for variable in range(start, stop, step):
    statements

```python
#Sum of Numbers
sum = 0

for i in range(1, 6):
    sum += i

print("Sum =", sum)
```

```python
#Print Even Numbers
for i in range(2, 11, 2):
    print(i)
```

```python
#Print Numbers 0 to 4
for i in range(5):
    print(i)
```

While Loop in Python


A while loop executes as long as the given condition is True.


Syntax

while condition:
    statements

```python
#Print Numbers 1 to 5

num = 1

while num <= 5:
    print(num)
    num += 1
```

```python
#Print Even Numbers

num = 2

while num <= 10:
    print(num)
    num += 2
```

```python
#Sum of Numbers 1 to 5
num = 1
sum = 0

while num <= 5:
    sum += num
    num += 1

print("Sum =", sum)
```

```python
#Even or Odd Number

num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
#Simple Linear Prediction (y = 2x + 3)

x = int(input("Enter x value: "))

y = 2 * x + 3

print("x =", x)
print("y =", y)
```

```python
#Average of 5 Numbers

a = int(input("Enter 1st number: "))
b = int(input("Enter 2nd number: "))
c = int(input("Enter 3rd number: "))
d = int(input("Enter 4th number: "))
e = int(input("Enter 5th number: "))

avg = (a + b + c + d + e) / 5

print("Average =", avg)
```

```python
#Temperature Classification

temp = int(input("Enter Temperature: "))

if temp > 30:
    print("Hot")
else:
    print("Normal")
```

```python
#Pattern

for i in range(5):
    for j in range(i + 1):
        print("*", end=" ")
    print()
```

```python
#Armstrong Number
num = 153
temp = num
sum = 0

n = len(str(num))

while num > 0:
    rem = num % 10
    sum = sum + rem ** n
    num = num // 10

if temp == sum:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
#Palindrome Number

num = 121
temp = num
reverse = 0

while num > 0:
    remainder = num % 10
    reverse = reverse * 10 + remainder
    num = num // 10

if temp == reverse:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
#Factorial of a Number

digit = 5
fact = 1

for i in range(1, digit + 1):
    fact = fact * i

print(fact)
```

```python
#Prime Number Check

num = int(input("Enter a number: "))
count = 0

for i in range(1, num + 1):
    if num % i == 0:
        count += 1

if count == 2:
    print("It is a Prime Number")
else:
    print("Not a Prime Number")
```

```python
#Nested If (Voting Eligibility)

age = int(input("Enter your age: "))
citizen = True

if citizen:
    if age >= 18:
        print("Eligible")
    else:
        print("Not Eligible")
```

```python
#Grade Calculation using if-elif-else

marks = int(input("Enter marks: "))

if 80 <= marks <= 100:
    print("A Grade")
elif 50 <= marks < 80:
    print("B Grade")
elif 40 <= marks < 50:
    print("C Grade")
elif 0 <= marks < 40:
    print("Fail")
else:
    print("Invalid Marks")
```
